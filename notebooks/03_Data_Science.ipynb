```python
# ============================================================
# DATA SCIENCE INPUT AUDIT
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("DATA SCIENCE INPUT AUDIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. Separate features and target
# ------------------------------------------------------------

feature_columns = [col for col in train_features.columns
                   if col.startswith("MFCC_")]

target_column = "emotion"

print(f"\nNumber of MFCC features: {len(feature_columns)}")
print(f"Target column: {target_column}")

print("\nFeature columns:")
print(feature_columns)

# ------------------------------------------------------------
# 2. Check data types
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE DATA TYPES")
print("=" * 70)

print(train_features[feature_columns].dtypes.value_counts())

# ------------------------------------------------------------
# 3. Missing values
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

train_missing = train_features[feature_columns].isna().sum().sum()
test_missing = test_features[feature_columns].isna().sum().sum()

print("Train missing values:", train_missing)
print("Test missing values:", test_missing)

# ------------------------------------------------------------
# 4. Infinite values
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("INFINITE VALUES")
print("=" * 70)

train_inf = np.isinf(
    train_features[feature_columns].to_numpy()
).sum()

test_inf = np.isinf(
    test_features[feature_columns].to_numpy()
).sum()

print("Train infinite values:", train_inf)
print("Test infinite values:", test_inf)

# ------------------------------------------------------------
# 5. Target distribution
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET DISTRIBUTION")
print("=" * 70)

print("\nTRAIN:")
print(train_features[target_column].value_counts())

print("\nTEST:")
print(test_features[target_column].value_counts())

# ------------------------------------------------------------
# 6. Unique classes
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("EMOTION CLASSES")
print("=" * 70)

train_classes = sorted(train_features[target_column].unique())
test_classes = sorted(test_features[target_column].unique())

print("Train classes:", train_classes)
print("Test classes:", test_classes)

print(
    "\nSame classes in train/test:",
    train_classes == test_classes
)

# ------------------------------------------------------------
# 7. Duplicate filenames
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FILENAME CHECK")
print("=" * 70)

train_duplicate_files = train_features["filename"].duplicated().sum()
test_duplicate_files = test_features["filename"].duplicated().sum()

print("Duplicate train filenames:", train_duplicate_files)
print("Duplicate test filenames:", test_duplicate_files)

# ------------------------------------------------------------
# 8. Train/test filename overlap
# ------------------------------------------------------------

train_files = set(train_features["filename"])
test_files = set(test_features["filename"])

overlap = train_files.intersection(test_files)

print("Train/Test filename overlap:", len(overlap))

# ------------------------------------------------------------
# 9. Feature statistics
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE STATISTICS")
print("=" * 70)

feature_summary = train_features[feature_columns].describe().T

display(feature_summary.head(10))

# ------------------------------------------------------------
# 10. Final audit result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("AUDIT SUMMARY")
print("=" * 70)

checks = {
    "40 MFCC features": len(feature_columns) == 40,
    "No train missing values": train_missing == 0,
    "No test missing values": test_missing == 0,
    "No train infinite values": train_inf == 0,
    "No test infinite values": test_inf == 0,
    "Same emotion classes": train_classes == test_classes,
    "No duplicate train filenames": train_duplicate_files == 0,
    "No duplicate test filenames": test_duplicate_files == 0,
    "No train/test filename overlap": len(overlap) == 0
}

for check, result in checks.items():
    print(f"{'PASS' if result else 'FAIL'} | {check}")
```

```python
# ============================================================
# DATA SCIENCE - STEP 2
# PREPARE TRAINING AND TEST DATA
# ============================================================

print("=" * 70)
print("PREPARING DATA FOR MODELING")
print("=" * 70)

# Separate features and target
X_train = train_features.drop(columns=["emotion", "filename"], errors="ignore")
y_train = train_features["emotion"]

X_test = test_features.drop(columns=["emotion", "filename"], errors="ignore")
y_test = test_features["emotion"]

print("\nTraining data:")
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("\nTest data:")
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

print("\nEmotion classes:")
print(sorted(y_train.unique()))

print("\nTraining class distribution:")
print(y_train.value_counts().sort_index())

print("\nTest class distribution:")
print(y_test.value_counts().sort_index())

print("\n" + "=" * 70)
print("DATA PREPARATION COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# DATA SCIENCE - STEP 3
# LABEL ENCODING + GROUP-AWARE VALIDATION SETUP
# ============================================================

import numpy as np
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import GroupKFold

print("=" * 70)
print("SETTING UP LABEL ENCODING AND LEAKAGE-SAFE VALIDATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. Encode emotion labels
# ------------------------------------------------------------

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("\nEmotion label mapping:")

for encoded_value, emotion in enumerate(label_encoder.classes_):
    print(f"{encoded_value} -> {emotion}")


# ------------------------------------------------------------
# 2. Obtain actor IDs for the training samples
# ------------------------------------------------------------

groups_train = train_metadata["actor_id"].values

print("\nNumber of training samples:", len(X_train))
print("Number of actor-group labels:", len(groups_train))

print("\nUnique training actors:")
print(sorted(np.unique(groups_train)))


# ------------------------------------------------------------
# 3. Verify that the test actors are completely separate
# ------------------------------------------------------------

train_actors = set(train_metadata["actor_id"])
test_actors = set(test_metadata["actor_id"])

actor_overlap = train_actors.intersection(test_actors)

print("\nTraining actors:", sorted(train_actors))
print("Testing actors:", sorted(test_actors))

print("\nActor overlap:", actor_overlap)

if len(actor_overlap) == 0:
    print("LEAKAGE CHECK: PASSED")
else:
    print("LEAKAGE CHECK: FAILED")


# ------------------------------------------------------------
# 4. Create GroupKFold validation strategy
# ------------------------------------------------------------

group_kfold = GroupKFold(n_splits=5)

print("\nGroupKFold configured:")
print("Number of folds:", group_kfold.n_splits)

# Verify each validation fold contains unseen actors
print("\nChecking actor separation across folds:")

for fold, (train_idx, val_idx) in enumerate(
    group_kfold.split(X_train, y_train_encoded, groups=groups_train),
    start=1
):

    fold_train_actors = set(groups_train[train_idx])
    fold_val_actors = set(groups_train[val_idx])

    overlap = fold_train_actors.intersection(fold_val_actors)

    print(
        f"Fold {fold}: "
        f"Train samples = {len(train_idx)}, "
        f"Validation samples = {len(val_idx)}, "
        f"Train actors = {len(fold_train_actors)}, "
        f"Validation actors = {len(fold_val_actors)}, "
        f"Actor overlap = {len(overlap)}"
    )

    if len(overlap) != 0:
        print("WARNING: Actor leakage detected!")

print("\n" + "=" * 70)
print("VALIDATION SETUP COMPLETE")
print("=" * 70)
```
